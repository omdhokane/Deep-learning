```python
from ultralytics import YOLO
import cv2
```

```python
model=YOLO("yolov8n.pt")
```

```python
cam=cv2.VideoCapture(1)
```

```python
while True:
    ret,frame=cam.read()
    result=model(frame)
    
    annotated_frame = result[0].plot()

    cv2.imshow("YOLO Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('x'):
        break
cam.release()
cv2.destroyAllWindows()
```
